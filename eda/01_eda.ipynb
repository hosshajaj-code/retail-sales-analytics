```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.barplot(data=df, x='Category', y='Profit', estimator=sum, palette='viridis')
plt.title('Total Profit per Category')
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, y='Profit', ax=axes[0], color='orange')
axes[0].set_title('Boxplot of Profit (Checking Outliers)')

sns.boxplot(data=df, y='Sales', ax=axes[1], color='skyblue')
axes[1].set_title('Boxplot of Sales (Checking Outliers)')

plt.show()
```
